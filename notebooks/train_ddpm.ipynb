# Using DDPM to generate MNIST digits conditionally

```python
import torch
import torch.nn.functional as F
from torch.utils.data import DataLoader

from torchvision import transforms
from torchvision.datasets import MNIST

import os
from tqdm import tqdm
import numpy as np
import matplotlib.pyplot as plt
from IPython.display import HTML

from diffusion_mnist.models import ContextUnet 
from diffusion_mnist.hyperparams import Hyperparameters
from diffusion_mnist import utils
```

# Denoising parameters schedule

DDPM proposes a forward process whose variance $\beta_t$ follows a schedule along the timesteps. So, to denoise the samples, the parameters also should be scheduled accordingly: 

* $ \beta_t = \lfloor \left(\frac{\Beta_2 - \Beta_1}{N}\right)t \rfloor + \Beta_1 $
* $ \alpha_t = 1 - \beta_t $
* $ \bar{\alpha}_t = \prod_{s=0}^t \alpha_s $
* $ \bar{\alpha}_0 = 1 $

Where $\Beta_1$ and $\Beta_2$ are, respectively, the forward process variance upper and lower bounds and N is the number of timesteps.

```python
# construct DDPM noise schedule
b_t = (Hyperparameters.beta2 - Hyperparameters.beta1) * torch.linspace(0, 1, Hyperparameters.timesteps + 1, device=Hyperparameters.DEVICE) + Hyperparameters.beta1
a_t = 1 - b_t
ab_t = torch.cumsum(a_t.log(), dim=0).exp()    
ab_t[0] = 1
```

# Remove and add noise

Consider the probability distribution of the reverse Markov chain process:

$$ p_\theta(x_{t-1}|x_t) = \mathcal{N}(x_{t-1}; \mu_\theta(x_t, t), \Sigma_\theta(x_t, t)) $$

DDPM relies on sampling iteratively from this distribution starting at $p(x_T) = \mathcal{N}(x_T; \mathbb{0}, \mathbb{I})$. The paper proposes that the distribution for the reverse process can be parameterized with the following mean and variance:

$$ \mu_\theta(x_t, t) = \frac{1}{\sqrt{\alpha_\theta}} \left(x_t - \frac{\beta_t}{\sqrt{1-\bar{\alpha}_t}} \epsilon_\theta (x_t, t) \right) $$
$$ \Sigma_\theta(x_t, t) = \sqrt{\beta_t}$$

Where $\epsilon_\theta$ is the Gaussian noise predicted by the model given a sample $x_t$ in timestep $t$. From this relationships, we can sample $x_{t-1}$ given a sample $x_t$ iteratively until we get to $x_0$ which is our generated data:

$$ x_{t-1}(x_t) = \mu_\theta(x_t, t) + \Sigma_\theta(x_t, t) $$

```python
# calculates x_{t-1} from x_t by removing the predicted noise and adding the scaled noise (to avoid collapse)
def denoise_add_noise(x, t, pred_noise, z=None):
    if z is None:
        z = torch.randn_like(x)
    noise = b_t.sqrt()[t] * z
    mean = (x - pred_noise * ((1 - a_t[t]) / (1 - ab_t[t]).sqrt())) / a_t[t].sqrt() # TO-DO: check if 1-a_t[t] can be replaced with b_t[t]
    return mean + noise
```

# Creating the model

```python
# construct model
model = ContextUnet(in_channels=1, n_feat=Hyperparameters.n_feat, n_cfeat=Hyperparameters.n_cfeat, height=Hyperparameters.height).to(Hyperparameters.DEVICE)
# optim = torch.optim.Adam(model.parameters(), lr=Hyperparameters.learning_rate)

# load in model weights and set to eval mode
model.load_state_dict(torch.load(f"weights/ddpm.pt", map_location=Hyperparameters.DEVICE))
# model.eval()
```

# Sampling iteratively from the Markov chain

Now we create a function to sample iteratively from $p(x_T) = \mathcal{N}(x_T; \mathbb{0}, \mathbb{I})$ until we get the generated data $x_0$. Since we are building a conditional diffusion model, the generated data depends on a context (the MNIST digit we want to generated) that should be given to the model.

```python
@torch.no_grad()
def sample_ddpm_context(n_sample, context, save_rate=20):
    # x_T ~ N(0, 1), sample initial noise
    samples = torch.randn(n_sample, Hyperparameters.n_channels, Hyperparameters.height, Hyperparameters.height).to(Hyperparameters.DEVICE)  

    # array to keep track of generated steps for plotting
    intermediate = [] 
    for i in range(Hyperparameters.timesteps, 0, -1):
        print(f'sampling timestep {i:3d}', end='\r')

        # reshape time tensor. time is represented by a linespace between 0 and 1 with N values.
        t = torch.tensor([i / Hyperparameters.timesteps])[:, None, None, None].to(Hyperparameters.DEVICE)

        # sample some random noise to inject back in. For i = 1, don't add back in noise
        z = torch.randn_like(samples) if i > 1 else 0

        eps = model(samples, t, c=context)    # predict noise e_(x_t,t, ctx)
        samples = denoise_add_noise(samples, i, eps, z)
        if i % save_rate==0 or i==Hyperparameters.timesteps or i<8:
            intermediate.append(samples.detach().cpu().numpy())

    intermediate = np.stack(intermediate)
    return samples, intermediate # final sample and intermediate latents
```

A notable property of the forward process is that it admits sampling $x_t$ at an arbitrary timestep $t$ in closed form:

$$ q(x_t|x_0) = \mathcal{N} (x_t; \sqrt{\bar{a}_t}x_0, (1 − \bar{a}_t)\mathbb{I})  $$

```python
# helper function: perturbs an image to a specified noise level
def perturb_input(x, t, noise):
    return ab_t.sqrt()[t, None, None, None] * x + (1 - ab_t[t, None, None, None]) * noise
```

# Loading data

```python
dataset = MNIST(root='data', download=True, transform=
        transforms.Compose([
        transforms.ToTensor(),                # from [0,255] to range [0.0,1.0]
        transforms.Normalize((0.5,), (0.5,))  # range [-1,1]
]))

dataloader = DataLoader(dataset, batch_size=Hyperparameters.batch_size, shuffle=True, num_workers=1)
```

# Training loop

```python
# training without context code

# set into train mode
model.train()

for ep in range(Hyperparameters.n_epochs):
    print(f'epoch {ep}')
    
    # linearly decay learning rate
    optim.param_groups[0]['lr'] = Hyperparameters.learning_rate*(1-ep/Hyperparameters.n_epochs)
    
    pbar = tqdm(dataloader, mininterval=2 )
    for x, c in pbar:   # x: images  c: context
        optim.zero_grad()
        x = x.to(Hyperparameters.DEVICE)
        c = F.one_hot(c, num_classes=10).float().to(Hyperparameters.DEVICE)

        # randomly mask out c
        context_mask = torch.bernoulli(torch.zeros(c.shape[0]) + 0.9).to(Hyperparameters.DEVICE)
        c = c * context_mask.unsqueeze(-1)
        
        # perturb data
        noise = torch.randn_like(x)
        t = torch.randint(1, Hyperparameters.timesteps + 1, (x.shape[0],)).to(Hyperparameters.DEVICE) 
        x_pert = perturb_input(x, t, noise)
        
        # use network to recover noise
        pred_noise = model(x_pert, t / Hyperparameters.timesteps, c=c)
        
        # loss is mean squared error between the predicted and true noise
        loss = F.mse_loss(pred_noise, noise)
        loss.backward()
        
        optim.step()

    # save model periodically
    if ep%4==0 or ep == int(Hyperparameters.n_epochs-1):
        if not os.path.exists(Hyperparameters.save_dir):
            os.mkdir(Hyperparameters.save_dir)
        torch.save(model.state_dict(), Hyperparameters.save_dir + f"context_model_{ep}.pth")
        print('saved model at ' + Hyperparameters.save_dir + f"context_model_{ep}.pth")
```
